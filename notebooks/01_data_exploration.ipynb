```python
import pandas as pd

# Load the articles dataset
df = pd.read_csv('../data/aging_narratives_articles.csv')
df_2 = pd.read_csv('../data/aging_narratives_articles_merged.csv')
df_missing = pd.read_csv('../data/aging_narratives_articles_missing_provinces.csv')


print(f"Total articles: {len(df)}")
print(f"\nColumns: {df.columns.tolist()}")

print(f"Total articles: {len(df_2)}")

```

```python
# Check missing values
print("Missing values:")
print(df_2.isnull().sum())

# Check article_text quality
lengths = df_2['article_text'].str.len()
print(f"\n--- Article Text Quality ---")
print(f"Empty/NaN: {df['article_text'].isna().sum()}")
print(f"1-100 chars: {((lengths > 0) & (lengths <= 100)).sum()}")
print(f"101-500 chars: {((lengths > 100) & (lengths <= 500)).sum()}")
print(f"500+ chars (usable): {(lengths > 500).sum()}")
```

```python
# Remove rows with empty article_text
df_2 = df.dropna(subset=['article_text'])
print(f"After removing empty: {len(df)} articles")

# Check sample of short articles (101-500 chars)
lengths = df_2['article_text'].str.len()
short_articles = df_2[(lengths > 100) & (lengths <= 500)]
print(f"\n--- Sample short articles (101-500 chars) ---")
print(short_articles[['title', 'article_text']].head(3).to_string())
```
