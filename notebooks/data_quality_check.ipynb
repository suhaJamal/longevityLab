This notebook performs data quality checks:
1. Load the dataset
2. Find missing data
3. Remove duplicate columns
4. Check and remove duplicate rows

## Step 3: Find Missing Data

```python
# Count missing values per column
missing_counts = df.isnull().sum()
missing_percent = (df.isnull().sum() / len(df) * 100).round(2)

missing_summary = pd.DataFrame({
    'Missing Count': missing_counts,
    'Missing %': missing_percent
})

print("Missing Data Summary:")
print("=" * 50)
print(missing_summary[missing_summary['Missing Count'] > 0])

if missing_summary['Missing Count'].sum() == 0:
    print("No missing data found!")
else:
    print(f"\nTotal cells with missing data: {missing_summary['Missing Count'].sum()}")
```

```python
# Visualize missing data
print("Missing Data Heatmap (by column):")
print("=" * 50)

for col in df.columns:
    missing = df[col].isnull().sum()
    total = len(df)
    pct = missing / total * 100
    bar = '█' * int(pct / 2) + '░' * (50 - int(pct / 2))
    print(f"{col[:25]:<25} | {bar} | {missing:>6} ({pct:.1f}%)")
```

## Step 4: Check and Remove Duplicate Columns

```python
# Find duplicate columns (columns with identical values)
def find_duplicate_columns(dataframe):
    """Find columns that have identical values"""
    duplicate_cols = []
    cols = dataframe.columns.tolist()
    
    for i in range(len(cols)):
        for j in range(i + 1, len(cols)):
            if dataframe[cols[i]].equals(dataframe[cols[j]]):
                duplicate_cols.append((cols[i], cols[j]))
    
    return duplicate_cols

duplicate_columns = find_duplicate_columns(df)

print("Duplicate Columns Check:")
print("=" * 50)

if len(duplicate_columns) > 0:
    print(f"Found {len(duplicate_columns)} duplicate column pairs:")
    for col1, col2 in duplicate_columns:
        print(f"  - '{col1}' is identical to '{col2}'")
else:
    print("No duplicate columns found!")
```

```python
# Check for columns with same names (if any)
column_names = df.columns.tolist()
duplicate_names = [col for col in column_names if column_names.count(col) > 1]

print("Duplicate Column Names Check:")
print("=" * 50)

if len(duplicate_names) > 0:
    print(f"Found columns with duplicate names: {set(duplicate_names)}")
else:
    print("No duplicate column names found!")
```

```python
# Remove specific columns
columns_to_remove = ['phase_category', 'province', 'row_hash', 'scrape_method']

# Drop columns (errors='ignore' in case column doesn't exist)
df_clean = df.drop(columns=columns_to_remove, errors='ignore')

print(f"Columns removed: {columns_to_remove}")
print(f"Shape after removal: {df_clean.shape}")
```

---
## Step 5: Check and Remove Duplicate Rows

```python
# Check duplicate rows based on URL
duplicate_urls = df_clean.duplicated(subset=['url']).sum()
print(f"Duplicate rows (based on URL): {duplicate_urls}")

# Remove duplicates based on URL (keep first occurrence)
rows_before = len(df_clean)
df_clean = df_clean.drop_duplicates(subset=['url'], keep='first')
rows_after = len(df_clean)

print(f"Rows before: {rows_before}")
print(f"Rows after: {rows_after}")
print(f"Rows removed: {rows_before - rows_after}")
```

```python
from openai import OpenAI

# API Key
API_KEY = "..."
client = OpenAI(api_key=API_KEY)

# Get rows with missing ai_label
missing_label = df_clean[df_clean['ai_label'].isna() | (df_clean['ai_label'] == '')]
print(f"Articles with missing ai_label: {len(missing_label)}")

# Function to predict sentiment
def predict_sentiment_batch(articles_batch):
    articles_text = ""
    for i, (_, row) in enumerate(articles_batch.iterrows()):
        title = str(row['title'])[:100] if pd.notna(row['title']) else "No title"
        text = str(row['article_text']) if pd.notna(row['article_text']) else ""
        snippet = " ".join(text.split()[:250])
        articles_text += f"\nArticle {i+1}:\nTitle: {title}\nText: {snippet}...\n---\n"

    prompt = f"""Classify each article's sentiment about aging/older adults:

- **EMPOWERING:** Positive portrayal (active, contributing, resilient seniors; opportunities in aging).
- **LIMITING:** Negative portrayal (burden, decline, crisis, vulnerability, age as problem).
- **NEUTRAL:** Factual, balanced, or mixed portrayal without clear positive/negative framing.

{articles_text}

For each article, respond with ONLY the label (one per line):
EMPOWERING, LIMITING, or NEUTRAL"""

    try:
        response = client.chat.completions.create(
            model="gpt-3.5-turbo",
            messages=[{"role": "user", "content": prompt}],
            temperature=0,
            max_tokens=500
        )
        result = response.choices[0].message.content.strip()
        labels = []
        for line in result.split('\n'):
            line = line.strip().upper()
            if 'EMPOWERING' in line:
                labels.append('empowering')
            elif 'LIMITING' in line:
                labels.append('limiting')
            elif 'NEUTRAL' in line:
                labels.append('neutral')
        
        while len(labels) < len(articles_batch):
            labels.append('neutral')
        return labels[:len(articles_batch)]
    except Exception as e:
        print(f"Error: {e}")
        return ['neutral'] * len(articles_batch)

# Process in batches of 10
batch_size = 10
missing_indices = missing_label.index.tolist()

for i in range(0, len(missing_indices), batch_size):
    batch_indices = missing_indices[i:i+batch_size]
    batch = df_clean.loc[batch_indices]
    
    labels = predict_sentiment_batch(batch)
    
    for j, idx in enumerate(batch_indices):
        df_clean.at[idx, 'ai_label'] = labels[j]
    
    print(f"Processed {min(i+batch_size, len(missing_indices))}/{len(missing_indices)}")

# Verify
remaining_missing = df_clean['ai_label'].isna().sum() + (df_clean['ai_label'] == '').sum()
print(f"\nDone! Remaining missing ai_label: {remaining_missing}")
print(f"Label distribution:\n{df_clean['ai_label'].value_counts()}")
```

```python
import pandas as pd
import time
import json
import os
from openai import OpenAI
from tenacity import retry, stop_after_attempt, wait_exponential, retry_if_exception_type
from typing import List, Tuple

class TopicClassifier:
    def __init__(self, api_key: str, model: str = "gpt-3.5-turbo-0125"):
        """Initialize classifier with rate limiting and retry logic."""
        self.client = OpenAI(api_key=api_key)
        self.model = model
        self.cache_file = "../newData/cleanData/topic_classifications_cache.json"
        self.classifications_cache = self._load_cache()
        
        # Define clear, non-overlapping topics with examples
        self.TOPICS = {
            "Healthcare_Medical": "Hospital care, medical treatments, doctors, diseases, prescriptions",
            "SeniorLiving_Care": "Nursing homes, assisted living, home care services, care facilities",
            "Retirement_Finances": "Pensions, retirement savings, RRSPs, financial planning for retirement",
            "Policy_Government": "Government programs (OAS, CPP), legislation, public policy affecting seniors",
            "Social_Community": "Senior centers, community events, social isolation, volunteer programs",
            "Ageism_Rights": "Age discrimination, elder abuse, legal rights, workplace ageism",
            "Technology_Innovation": "Tech for seniors, telehealth, assistive devices, digital literacy",
            "Longevity_Research": "Aging science research, lifespan extension, anti-aging studies",
            "Housing_Accessibility": "Senior housing, accessibility modifications, affordable housing",
            "Family_Caregiving": "Family caregivers, intergenerational support, caregiver stress"
        }
    
    def _load_cache(self) -> dict:
        """Load previously classified articles from cache."""
        if os.path.exists(self.cache_file):
            try:
                with open(self.cache_file, 'r') as f:
                    return json.load(f)
            except:
                return {}
        return {}
    
    def _save_cache(self):
        """Save classifications to cache file."""
        with open(self.cache_file, 'w') as f:
            json.dump(self.classifications_cache, f)
    
    def _create_efficient_prompt(self, articles_batch: List[Tuple[str, str]]) -> str:
        """Create an optimized prompt that minimizes tokens while maintaining accuracy."""
        
        topics_list = "\n".join([f"{code}: {desc}" for code, desc in self.TOPICS.items()])
        
        articles_text = ""
        for i, (title, snippet) in enumerate(articles_batch):
            articles_text += f"{i+1}. Title: {title}\n   Content: {snippet}\n"
        
        prompt = f"""Classify each article into exactly ONE topic. Use EXACT topic codes.

TOPICS:
{topics_list}

ARTICLES:
{articles_text}

INSTRUCTIONS:
1. Analyze the MAIN focus of each article
2. Choose the SINGLE most appropriate topic code
3. Output ONLY the topic codes, one per line, in order

RESPONSE FORMAT:
TopicCode
TopicCode
TopicCode"""
        
        return prompt
    
    def _preprocess_article(self, text: str, max_words: int = 120) -> str:
        """Extract the most relevant part of the article."""
        if pd.isna(text):
            return ""
        
        text = str(text)
        words = text.split()
        
        if len(words) > 100:
            relevant = " ".join(words[:60] + words[-60:])
        else:
            relevant = " ".join(words[:max_words])
        
        return relevant
    
    @retry(
        stop=stop_after_attempt(5),
        wait=wait_exponential(multiplier=1, min=4, max=60),
        retry=retry_if_exception_type(Exception)
    )
    def _call_openai(self, prompt: str) -> List[str]:
        """Call OpenAI API with robust error handling."""
        try:
            response = self.client.chat.completions.create(
                model=self.model,
                messages=[
                    {"role": "system", "content": "You are a precise article classifier. Always output exactly the requested format."},
                    {"role": "user", "content": prompt}
                ],
                temperature=0.1,
                max_tokens=150,
                timeout=30
            )
            
            result = response.choices[0].message.content.strip()
            topics = [line.strip() for line in result.split('\n') if line.strip()]
            
            validated_topics = []
            for topic in topics:
                if topic in self.TOPICS:
                    validated_topics.append(topic)
                else:
                    matched = False
                    for valid_topic in self.TOPICS:
                        if valid_topic in topic or topic in valid_topic:
                            validated_topics.append(valid_topic)
                            matched = True
                            break
                    if not matched:
                        validated_topics.append("Unclassified")
            
            return validated_topics
            
        except Exception as e:
            print(f"API Error: {e}")
            raise
    
    def classify_batch(self, df_batch: pd.DataFrame) -> List[str]:
        """Classify a batch of articles efficiently."""
        
        articles_data = []
        for _, row in df_batch.iterrows():
            title = str(row['title'])[:60] if pd.notna(row.get('title')) else ""
            text = self._preprocess_article(row.get('article_text', ''), max_words=120)
            
            cache_key = f"{title}_{hash(text) % 100000}"
            if cache_key in self.classifications_cache:
                articles_data.append((title, "CACHED", self.classifications_cache[cache_key], cache_key))
            else:
                articles_data.append((title, text, None, cache_key))
        
        uncached_articles = [(title, text) for title, text, topic, _ in articles_data if topic is None]
        
        if not uncached_articles:
            return [topic for _, _, topic, _ in articles_data]
        
        prompt = self._create_efficient_prompt(uncached_articles)
        new_topics = self._call_openai(prompt)
        
        result_topics = []
        uncached_idx = 0
        
        for title, text, cached_topic, cache_key in articles_data:
            if cached_topic is not None:
                result_topics.append(cached_topic)
            else:
                topic = new_topics[uncached_idx] if uncached_idx < len(new_topics) else "Unclassified"
                result_topics.append(topic)
                self.classifications_cache[cache_key] = topic
                uncached_idx += 1
        
        if len(self.classifications_cache) % 50 == 0:
            self._save_cache()
        
        return result_topics
    
    def classify_dataframe(self, df: pd.DataFrame, batch_size: int = 8, save_interval: int = 100) -> pd.DataFrame:
        """Classify all articles in a dataframe with progress tracking."""
        
        print(f"Starting classification of {len(df)} articles...")
        print(f"Estimated cost: ${len(df) * 0.0001:.2f} - ${len(df) * 0.0002:.2f}")
        print(f"Estimated time: {len(df) / (batch_size * 2):.1f} minutes")
        print("-" * 50)
        
        all_topics = []
        start_time = time.time()
        
        for i in range(0, len(df), batch_size):
            batch = df.iloc[i:i+batch_size]
            
            try:
                batch_topics = self.classify_batch(batch)
                all_topics.extend(batch_topics)
                
                processed = min(i + batch_size, len(df))
                if processed % save_interval == 0 or processed == len(df):
                    elapsed = time.time() - start_time
                    rate = processed / elapsed if elapsed > 0 else 0
                    remaining = (len(df) - processed) / rate if rate > 0 else 0
                    
                    print(f"  Processed {processed}/{len(df)} | "
                          f"Rate: {rate:.1f} articles/sec | "
                          f"ETA: {remaining/60:.1f} min")
                    self._save_cache()
                
                time.sleep(0.2)
                
            except Exception as e:
                print(f"Error processing batch at index {i}: {e}")
                print("Waiting 10 seconds before retry...")
                time.sleep(10)
                
                try:
                    smaller_batch = df.iloc[i:i+max(1, batch_size//2)]
                    batch_topics = self.classify_batch(smaller_batch)
                    all_topics.extend(batch_topics)
                    print(f"  Successfully processed smaller batch")
                except:
                    all_topics.extend(["Failed"] * len(batch))
                    print(f"  Marked batch as failed")
        
        df = df.copy()
        df['topic'] = all_topics[:len(df)]
        
        self._save_cache()
        
        print("\n" + "=" * 50)
        print("CLASSIFICATION COMPLETE!")
        print("=" * 50)
        print(f"\nTopic distribution:")
        print(df['topic'].value_counts())
        
        return df


# ============================================
# RUN CLASSIFICATION
# ============================================

# API Key
API_KEY = "..."

# Initialize classifier
classifier = TopicClassifier(api_key=API_KEY)

# Classify articles
df_clean = classifier.classify_dataframe(df_clean, batch_size=8, save_interval=100)

# Show results
print("\nSample of classified articles:")
print(df_clean[['title', 'topic']].head(10))
```
